# 2. Introduction to the Pre-trained Model

YOLO11 is a deep learning model used for real-time object detection.

YOLO stands for You Only Look Once.

YOLO11n is the nano version of the YOLO11 model family. It is lightweight and suitable for applications requiring faster inference.

The model is initially trained on a general object detection dataset. In this project, the pre-trained model is adapted to detect helmet and no-helmet classes.

### Model Information

- Model Name: YOLO11n
- Model Type: Object Detection
- Pre-trained Weights: yolo11n.pt
- Input: Images and video frames
- Output: Bounding boxes, class labels, and confidence scores
- Application: Helmet Detection

##  Model Architecture and Working Principle

YOLO11 is a one-stage object detection model that identifies objects and predicts their locations in an image in a single forward pass.

The model produces:

* Bounding box coordinates for detected objects.
* Class labels such as helmet and no_helmet.
* Confidence scores indicating the model's confidence in each detection.

### Workflow

Input Image → Preprocessing → Pre-trained YOLO11 Model → Inference → Post-processing → Bounding Boxes and Class Labels → Final Output

The input image is resized and prepared for the model. YOLO11 predicts the objects, and the output is processed using confidence thresholds and non-maximum suppression to remove overlapping detections.


## Why This Model Was Selected

YOLO11 was selected because it supports real-time object detection and can identify multiple objects in an image. It is suitable for helmet detection because it can predict object locations, class labels, and confidence scores.

The project uses the pretrained YOLO11n model and fine-tunes it on the helmet detection dataset. This approach uses previously learned features instead of training a deep learning model completely from scratch.

```python
import os
from glob import glob

source = "/content/helmet_dataset"

all_xml = glob(
    source + "/**/*.xml",
    recursive=True
)

all_images = []

for ext in ["jpg", "jpeg", "png"]:
    all_images.extend(
        glob(source + f"/**/*.{ext}", recursive=True)
    )

print("Total XML files:", len(all_xml))
print("Total images:", len(all_images))

print("Sample XML files:", all_xml[:5])
print("Sample image files:", all_images[:5])
```

```python
import xml.etree.ElementTree as ET
from collections import Counter

class_counts = Counter()

for xml_file in all_xml:
    root = ET.parse(xml_file).getroot()

    for obj in root.findall("object"):
        name = obj.findtext("name", "").strip().lower()
        class_counts[name] += 1

print("Classes found in dataset:")

for name, count in class_counts.items():
    print(name, ":", count)
```

```python
import os
import random
import shutil
import xml.etree.ElementTree as ET
from glob import glob

random.seed(42)

source = "/content/helmet_dataset"
output = "/content/helmet_yolo"

for split in ["train", "val"]:
    os.makedirs(f"{output}/images/{split}", exist_ok=True)
    os.makedirs(f"{output}/labels/{split}", exist_ok=True)

image_files = []

for ext in ["jpg", "jpeg", "png"]:
    image_files.extend(
        glob(source + f"/**/*.{ext}", recursive=True)
    )

xml_files = glob(source + "/**/*.xml", recursive=True)

xml_map = {
    os.path.splitext(os.path.basename(f))[0]: f
    for f in xml_files
}

pairs = []

for img in image_files:
    name = os.path.splitext(os.path.basename(img))[0]

    if name in xml_map:
        pairs.append((img, xml_map[name]))

random.shuffle(pairs)

split_index = int(len(pairs) * 0.8)

train_pairs = pairs[:split_index]
val_pairs = pairs[split_index:]

print("Training images:", len(train_pairs))
print("Validation images:", len(val_pairs))

class_map = {
    "with helmet": 0,
    "helmet": 0,
    "with_helmet": 0,
    "without helmet": 1,
    "no helmet": 1,
    "no_helmet": 1,
    "without_helmet": 1
}

def convert_xml(xml_path, label_path):
    root = ET.parse(xml_path).getroot()

    size = root.find("size")
    img_width = int(size.findtext("width"))
    img_height = int(size.findtext("height"))

    lines = []

    for obj in root.findall("object"):
        name = obj.findtext("name", "").strip().lower()

        if name not in class_map:
            continue

        class_id = class_map[name]

        box = obj.find("bndbox")

        xmin = float(box.findtext("xmin"))
        ymin = float(box.findtext("ymin"))
        xmax = float(box.findtext("xmax"))
        ymax = float(box.findtext("ymax"))

        x_center = ((xmin + xmax) / 2) / img_width
        y_center = ((ymin + ymax) / 2) / img_height

        width = (xmax - xmin) / img_width
        height = (ymax - ymin) / img_height

        lines.append(
            f"{class_id} {x_center} {y_center} {width} {height}"
        )

    with open(label_path, "w") as f:
        f.write("\n".join(lines))

def prepare_split(data, split):
    for img_path, xml_path in data:
        filename = os.path.basename(img_path)
        name = os.path.splitext(filename)[0]

        new_img = f"{output}/images/{split}/{filename}"
        new_label = f"{output}/labels/{split}/{name}.txt"

        shutil.copy2(img_path, new_img)
        convert_xml(xml_path, new_label)

prepare_split(train_pairs, "train")
prepare_split(val_pairs, "val")

print("YOLO conversion completed!")
```

```python
yaml_content = """
path: /content/helmet_yolo

train: images/train
val: images/val

names:
  0: helmet
  1: no_helmet
"""

with open("/content/helmet_yolo/data.yaml", "w") as f:
    f.write(yaml_content)

print("data.yaml created successfully")
```

```python
from ultralytics import YOLO

model = YOLO("yolo11n.pt")

print("YOLO11 model loaded successfully")
```

```python
results = model.train(
    data="/content/helmet_yolo/data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,
    device=0,
    patience=10,
    workers=2,
    project="/content/runs",
    name="helmet_detection"
)
```

```python
from ultralytics import YOLO

best_model = YOLO(
    "/content/runs/helmet_detection/weights/best.pt"
)

metrics = best_model.val(
    data="/content/helmet_yolo/data.yaml",
    imgsz=640
)

print("mAP50:", metrics.box.map50)
print("mAP50-95:", metrics.box.map)
print("Precision:", metrics.box.mp)
print("Recall:", metrics.box.mr)
```
